```python
from agentsq.agent.tools.mcp import McpTools
```

```python
async with McpTools() as backend:
    specs = await backend.specs()
    for s in specs:
        props = s["parameters"]["properties"]
        required = s["parameters"].get("required", [])
        print(s["name"])
        print("  required:", required)
        print("  all keys:", list(props))
```
